# NumPy Statistical Analysis

This notebook analyzes a numerical **Student Performance Dataset** using NumPy statistical functions.


## Objective
Practice applying statistical functions to numerical data and identify important patterns.

```python
import numpy as np

# Student marks out of 100
marks = np.array([45, 52, 58, 61, 64, 67, 70, 72, 75, 78, 82, 85, 90, 95, 98])

print("Dataset:", marks)
```

## Statistical Calculations

```python
mean = np.mean(marks)
median = np.median(marks)
std_dev = np.std(marks)
minimum = np.min(marks)
maximum = np.max(marks)

print(f"Mean: {mean:.2f}")
print(f"Median: {median:.2f}")
print(f"Standard Deviation: {std_dev:.2f}")
print(f"Minimum: {minimum}")
print(f"Maximum: {maximum}")
```

## Compare Mean and Median

```python
difference = mean - median
print(f"Mean = {mean:.2f}")
print(f"Median = {median:.2f}")
print(f"Difference = {difference:.2f}")

if mean > median:
    print("The mean is slightly higher than the median, suggesting a small influence from higher scores.")
elif mean < median:
    print("The mean is slightly lower than the median, suggesting a small influence from lower scores.")
else:
    print("The mean and median are equal.")
```

## Check for Unusually High or Low Values

```python
low_threshold = mean - 2 * std_dev
high_threshold = mean + 2 * std_dev

unusual_values = marks[(marks < low_threshold) | (marks > high_threshold)]

print(f"Lower threshold: {low_threshold:.2f}")
print(f"Upper threshold: {high_threshold:.2f}")
print("Unusually low/high values:", unusual_values if len(unusual_values) else "None")
```
